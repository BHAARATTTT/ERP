```python
import pandas as pd

# Load datasets
holdings_df = pd.read_csv("holdings.csv")
trades_df = pd.read_csv("trades.csv")

print("Holdings Data:")
display(holdings_df.head())

print("Trades Data:")
display(trades_df.head())
```

```python
print("Holdings columns:")
print(holdings_df.columns)

print("\nTrades columns:")
print(trades_df.columns)
```

```python
# --- Holdings count per fund ---
holdings_count = (
    holdings_df
    .groupby("PortfolioName")
    .size()
    .reset_index(name="TotalHoldings")
)

# --- Trades count per fund ---
trades_count = (
    trades_df
    .groupby("PortfolioName")
    .size()
    .reset_index(name="TotalTrades")
)

# --- Yearly P&L per fund ---
yearly_pnl = (
    holdings_df
    .groupby("PortfolioName")[["PL_YTD"]]
    .sum()
    .reset_index()
)

display(holdings_count.head())
display(trades_count.head())
display(yearly_pnl.head())
```

```python
def chatbot(query):
    query = query.lower()

    # --- Total holdings for a fund ---
    if "holding" in query:
        for fund in holdings_count["PortfolioName"]:
            if fund.lower() in query:
                total = holdings_count[
                    holdings_count["PortfolioName"] == fund
                ]["TotalHoldings"].values
                if len(total) > 0:
                    return f"Total number of holdings for {fund} is {int(total[0])}"

    # --- Total trades for a fund ---
    if "trade" in query:
        for fund in trades_count["PortfolioName"]:
            if fund.lower() in query:
                total = trades_count[
                    trades_count["PortfolioName"] == fund
                ]["TotalTrades"].values
                if len(total) > 0:
                    return f"Total number of trades for {fund} is {int(total[0])}"

    # --- Best performing fund based on YTD P&L ---
    if "perform" in query or "best" in query:
        best_fund = yearly_pnl.sort_values("PL_YTD", ascending=False).iloc[0]
        return (
            f"The best performing fund is {best_fund['PortfolioName']} "
            f"with YTD P&L of {best_fund['PL_YTD']}"
        )

    # --- Fallback ---
    return "Sorry can not find the answer"
```

```python
# ---- Test queries ----

print(chatbot("Total number of holdings for Garfield"))
print(chatbot("How many trades does Account A have?"))
print(chatbot("Which fund performed best this year?"))

# ---- Negative test (important) ----
print(chatbot("What is the weather today?"))
```
